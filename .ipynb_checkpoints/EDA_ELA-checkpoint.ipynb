```python
df["search_score"] = (
    df["search_total"] -
    df["search_total"].min()
) / (
    df["search_total"].max() -
    df["search_total"].min()
)
```

```python
df["ctr_score"] = (
    df["ctr_mobile"] -
    df["ctr_mobile"].min()
) / (
    df["ctr_mobile"].max() -
    df["ctr_mobile"].min()
)
```

```python
competition_map = {
    "낮음": 1,
    "중간": 2,
    "높음": 3
}

df["competition_score"] = (
    df["competition"]
    .map(competition_map)
)

df["competition_opportunity"] = (
    4 - df["competition_score"]
) / 3
```

Opportunity Score는 NAVER 공식 지표가 아니다. 본 실습에서는 검색수요 50%, 클릭반응 30%, 낮은 경쟁도 20%를 반영하여 콘텐츠 기회를 탐색하기 위한 교육용 지표로 정의하였다.

```python
df["opportunity_score"] = (
    df["search_score"] * 0.5 +
    df["ctr_score"] * 0.3 +
    df["competition_opportunity"] * 0.2
)
```

```python
from collections import Counter
from kiwipiepy import Kiwi

kiwi = Kiwi()

word_list = []

for keyword in df["keyword"].dropna():

    tokens = kiwi.tokenize(str(keyword))

    for token in tokens:

        if token.tag in ["NNG", "NNP"]:
            if len(token.form) >= 2:
                word_list.append(token.form)

word_counts = Counter(word_list)

words = [
    word
    for word, count
    in word_counts.most_common(20)
]

words
```

```python
word_counts = []

for word in words:
    
    count = (
        df["keyword"]
        .str.contains(
            word,
            case=False,
            na=False
        )
        .sum()
    )
    
    word_counts.append({
        "word": word,
        "count": count
    })

word_df = pd.DataFrame(word_counts)
```

```python
intent_keywords = {

    # 1. 인기/순위를 확인하려는 검색
    "Ranking": [
        "베스트",
        "베스트셀러",
        "순위",
        "인기",
        "랭킹"
    ],

    # 2. 어떤 책을 선택할지 추천받으려는 검색
    "Recommendation": [
        "추천",
        "추천책",
        "책추천",
        "도서추천",
        "추천도서"
    ],

    # 3. 책을 구매하거나 혜택을 확인하려는 검색
    "Purchase": [
        "구매",
        "가격",
        "할인",
        "쿠폰",
        "적립",
        "배송",
        "새벽배송"
    ],

    # 4. 책을 읽거나 이용할 방법을 찾는 검색
    "Access": [
        "도서관",
        "대여",
        "전자책",
        "ebook",
        "e북",
        "구독",
        "무료"
    ],

    # 5. 책이나 독서에 대한 정보를 찾는 검색
    "Information": [
        "정보",
        "소개",
        "뜻",
        "방법",
        "독서",
        "서평",
        "리뷰",
        "목록"
    ]
}
```

```python
def classify_intent(keyword):
    
    keyword = str(keyword)
    
    for intent, words in intent_keywords.items():
        
        for word in words:
            
            if word.lower() in keyword.lower():
                return intent
    
    return "Others"
```

```python
df["intent"] = (
    df["keyword"]
    .apply(classify_intent)
)
```

```python
seed_df = (
    df[
        df["intent"] != "Others"
    ]
    .dropna(
        subset=[
            "keyword",
            "opportunity_score",
            "intent"
        ]
    )
    .sort_values(
        "opportunity_score",
        ascending=False
    )
    .groupby(
        "intent",
        as_index=False
    )
    .first()
    .sort_values(
        "opportunity_score",
        ascending=False
    )
)

seed_df[
    [
        "intent",
        "keyword",
        "search_total",
        "ctr_mobile",
        "competition",
        "opportunity_score"
    ]
]
```

```python
def collect_blog_posts(
    query,
    target_count=200,
    sort="sim"
):
    
    posts = []
    
    # 검색 API pagination 범위를 고려
    max_count = min(target_count, 1000)
    
    for start in range(
        1,
        max_count + 1,
        100
    ):
        
        display = min(
            100,
            max_count - len(posts)
        )
        
        params = {
            "query": query,
            "display": display,
            "start": start,
            "sort": sort
        }
        
        response = requests.get(
            BLOG_API_URL,
            headers=headers,
            params=params,
            timeout=10
        )
        
        response.raise_for_status()
        
        data = response.json()
        items = data.get("items", [])
        
        if not items:
            break
        
        for item in items:
            
            posts.append({
                "query": query,
                "title": item.get("title"),
                "description": item.get("description"),
                "bloggername": item.get("bloggername"),
                "postdate": item.get("postdate"),
                "link": item.get("link")
            })
        
        if len(posts) >= max_count:
            break
    
    return posts
```

```python
# seed_keywords = [
#     "경제책",
#     "경제책추천",
#     "재테크책",
#     "투자책",
#     "주식책"
# ]

seed_keywords = (
    seed_df["keyword"]
    .tolist()
)

seed_keywords
```

```python
all_posts = []

for keyword in seed_keywords:
    
    print(f"{keyword} 수집 중...")
    
    posts = collect_blog_posts(
        query=keyword,
        target_count=100
    )
    
    print(
        f"→ {len(posts)}건 수집 완료"
    )
    
    all_posts.extend(posts)
```

```python
def clean_text(text):
    
    if pd.isna(text):
        return ""
    
    text = html.unescape(
        str(text)
    )
    
    # HTML 태그 제거
    text = re.sub(
        r"<[^>]+>",
        " ",
        text
    )
    
    # URL 제거
    text = re.sub(
        r"https?://\S+",
        " ",
        text
    )
    
    # 특수문자 제거
    text = re.sub(
        r"[^가-힣a-zA-Z0-9\s]",
        " ",
        text
    )
    
    # 연속 공백 제거
    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()
    
    return text
```

```python
noun_tags = {
    "NNG",
    "NNP"
}
```

```python
def extract_nouns(text):
    
    tokens = kiwi.tokenize(
        text
    )
    
    nouns = [
        token.form
        
        for token in tokens
        
        if token.tag in noun_tags
    ]
    
    return nouns
```

```python
stopwords = {
    "책",
    "도서",
    "추천",
    "소개",
    "내용",
    "관련",
    "이번",
    "오늘",
    "정도",
    "생각",
    "사람",
    "경우",
    "부분",
    "저자",
    "출판",
    "출판사",
    "블로그"
}
```

```python
def preprocess(text):
    
    nouns = extract_nouns(
        text
    )
    
    nouns = [
        noun
        
        for noun in nouns
        
        if (
            noun not in stopwords
            and len(noun) >= 2
        )
    ]
    
    return nouns
```

```python
def display_topics(
    model,
    feature_names,
    top_n=10
):
    
    topics = []
    
    for topic_idx, topic in enumerate(
        model.components_
    ):
        
        top_indices = (
            topic
            .argsort()[-top_n:]
            [::-1]
        )
        
        top_words = [
            feature_names[i]
            for i in top_indices
        ]
        
        topics.append({
            "topic": f"Topic {topic_idx + 1}",
            "keywords": ", ".join(top_words)
        })
    
    return pd.DataFrame(
        topics
    )
```

```python
topic_change = []

for topic in trend_df["topic"].unique():

    temp = (
        trend_df[
            trend_df["topic"] == topic
        ]
        .sort_values("period")
    )

    first_ratio = temp.iloc[0]["ratio"]
    last_ratio = temp.iloc[-1]["ratio"]

    topic_change.append({
        "topic": topic,
        "first_ratio": first_ratio,
        "last_ratio": last_ratio,
        "change": last_ratio - first_ratio
    })

change_df = pd.DataFrame(topic_change)

change_df.sort_values(
    "change",
    ascending=False
)
```

```python
fig = px.bar(
    change_df.sort_values("change"),
    x="change",
    y="topic",
    orientation="h",

    title="최근 1년 LDA 기반 Topic별 검색 관심도 변화",

    labels={
        "change": "관심도 변화",
        "topic": "LDA Topic"
    }
)

fig.show()
```
