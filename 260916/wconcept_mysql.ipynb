```python
# dictionary comprehension => 딕셔너리 라는 자료구조를 손쉽게 만들 수 있도록 해주는 문법체계

brand_map = {brand_name: brand_id for brand_id, brand_name in brand_rows}
brand_map
```

```python
PRODUCT_SQL = """
INSERT INTO products (
    product_id,
    brand_id,
    product_name,
    product_url,
    img_url,
    category_depth1,
    category_depth2,
    category_depth3,
    category_depth4
)
VALUES (
    %s, %s, %s, %s, %s,
    %s, %s, %s, %s
)
ON DUPLICATE KEY UPDATE
    brand_id = VALUES(brand_id),
    product_name = VALUES(product_name),
    product_url = VALUES(product_url),
    img_url = VALUES(img_url),
    category_depth1 = VALUES(category_depth1),
    category_depth2 = VALUES(category_depth2),
    category_depth3 = VALUES(category_depth3),
    category_depth4 = VALUES(category_depth4)
"""
```

```python
# 유틸리티 함수를 생성 & 사용
# 기존에 크롤링을 통해서 수집한 데이터 가운데 비어있는 = 결측치값이 존재하는 경우를 대비한 함수
# 파이썬 > 팬더스 > 결측치의 값이 NULL -> NaN, NaT // MySQL : None
import pandas as pd

def to_none(value) :
    if pd.isna(value) :
        return None
    return value

def to_date(value) :
    if pd.isna(value) :
        return None

    if isinstance(value, pd.Timestamp) :
        return value.date()
    return value
```

```python
for _, row in product_df.iterrows() :
    product_id = str(row["product_id"])
    brand_id = brand_map.get(row["brand_name"])

    cursor.execute(
        PRODUCT_SQL,
        (
            product_id,
            brand_id,
            to_none(row["product_name"]),
            to_none(row["product_url"]),
            to_none(row["img_url"]),
            to_none(row["category_depth1"]),
            to_none(row["category_depth2"]),
            to_none(row["category_depth3"]),
            to_none(row["category_depth4"])            
        )
    )
```

```python
SNAPSHOT_SQL = """
INSERT INTO product_snapshots (
    crawl_run_id,
    product_id,
    original_price,
    sale_price,
    discount_rate,
    rating,
    review_count,
    like_count
)
VALUES (
    %s, %s, %s, %s,
    %s, %s, %s, %s
)
ON DUPLICATE KEY UPDATE
    original_price = VALUES(original_price),
    sale_price = VALUES(sale_price),
    discount_rate = VALUES(discount_rate),
    rating = VALUES(rating),
    review_count = VALUES(review_count),
    like_count = VALUES(like_count)
"""
```

```python
for _, row in product_df.iterrows() :
    cursor.execute(
        SNAPSHOT_SQL,
        (
            crawl_run_id,
            str(row["product_id"]),
            to_none(row["original_price"]),
            to_none(row["sale_price"]),
            to_none(row["discount_rate"]),
            to_none(row["rating"]),
            to_none(row["review_count"]),
            to_none(row["like_count"])
        )
    )
```

```python
RANKING_SQL = """
INSERT INTO product_rankings (
    crawl_run_id,
    product_id,
    ranking_no
)
VALUES (
    %s, %s, %s
)
ON DUPLICATE KEY UPDATE
    ranking_no = VALUES(ranking_no)
"""
```

```python
for _, row in product_df.iterrows() :
    cursor.execute(
        RANKING_SQL,
        (
            crawl_run_id,
            str(row["product_id"]),
            int(row["ranking_no"])
        )
    )
```

```python
REVIEW_SQL = """
INSERT INTO reviews (
    crawl_run_id,
    product_id,
    review_page,
    review_no,
    reviewer,
    review_date,
    purchase_option,
    review_text,
    review_rating
)
VALUES (
    %s, %s, %s, %s, %s,
    %s, %s, %s, %s
)
ON DUPLICATE KEY UPDATE
    review_id = LAST_INSERT_ID(review_id),
    review_page = VALUES(review_page),
    review_no = VALUES(review_no),
    reviewer = VALUES(reviewer),
    review_date = VALUES(review_date),
    purchase_option = VALUES(purchase_option),
    review_text = VALUES(review_text),
    review_rating = VALUES(review_rating)
"""
```

```python
EVALUATION_SQL = """
INSERT INTO review_evaluations (
    review_id,
    evaluation_type,
    evaluation_value
)
VALUES (
    %s, %s, %s
)
ON DUPLICATE KEY UPDATE
    evaluation_type = VALUES(evaluation_type),
    evaluation_value = VALUES(evaluation_value)
"""
```

```python
IMAGE_SQL = """
INSERT INTO review_images (
    review_id,
    image_url
)
VALUES (
    %s, %s
)
ON DUPLICATE KEY UPDATE
    image_url = VALUES(image_url)
"""
```

```python
print(type(review_df.loc[0, "evaluations"]))
print(type(review_df.loc[0, "review_images"]))
```

```python
# Abstract Syntax Tree = 추상 구문 트리
# {"사이즈": "적당함", "색상": "같음", "소재": "같음"}	["https://media.wconcept.co.kr/review/30888695...

# {"사이즈": -> 딕셔너리 자료구조 , 키
# "적당함", -> 값
# "색상":
# "같음",
# "소재":
# "같음"}

import pandas as pd
import ast

review_df["evaluations"] = review_df["evaluations"].apply(
    lambda x: ast.literal_eval(x)
    if pd.notna(x)
    else {}
)

review_df["review_images"] = review_df["review_images"].apply(
    lambda x: ast.literal_eval(x)
    if pd.notna(x)
    else []
)
```

```python
print(type(review_df.loc[0, "evaluations"]))
print(type(review_df.loc[0, "review_images"]))
```

```python
for _, row in review_df.iterrows() :
    cursor.execute(
        REVIEW_SQL,
        (
            crawl_run_id,
            str(row["product_id"]),
            int(row["review_page"]),
            int(row["review_no"]),
            to_none(row["reviewer"]),
            to_date(row["review_date"]),
            to_none(row["purchase_option"]),
            to_none(row["review_text"]),
            to_none(row["review_rating"])
        )
    )

    review_id = cursor.lastrowid

    evaluations = row["evaluations"]

    if isinstance(evaluations, dict) :
        for evaluation_type, evaluation_value in evaluations.items() :
            cursor.execute(
                EVALUATION_SQL,
                (
                    review_id,
                    evaluation_type,
                    evaluation_value
                )
            )
            
    review_images = row["review_images"]

    if isinstance(review_images, list) :
        for image_url in review_images :

            if not image_url :
                continue
                
            cursor.execute(
                IMAGE_SQL,
                (
                    review_id,
                    image_url
                )
            )
```

```python
BLOG_SQL = """
INSERT INTO blog_posts (
    crawl_run_id,
    brand_id,
    search_brand,
    title,
    link,
    description,
    bloggername,
    bloggerlink,
    postdate
)
VALUES (
    %s, %s, %s, %s, %s,
    %s, %s, %s, %s
)
ON DUPLICATE KEY UPDATE
    title = VALUES(title),
    link = VALUES(link),
    description = VALUES(description),
    bloggername = VALUES(bloggername),
    bloggerlink = VALUES(bloggerlink),
    postdate = VALUES(postdate)
"""
```

```python
for _, row in blog_df.iterrows() :
    brand_id = brand_map.get(row["search_brand"])

    if brand_id is None :
        continue

    cursor.execute(
        BLOG_SQL,
        (
            crawl_run_id,
            brand_id,
            to_none(row["search_brand"]),
            to_none(row["title"]),
            to_none(row["link"]),
            to_none(row["description"]),
            to_none(row["bloggername"]),
            to_none(row["bloggerlink"]),
            to_date(row["postdate"])
        )
    )
```
