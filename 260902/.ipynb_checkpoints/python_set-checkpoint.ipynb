### 변수 & 자료형
1. 기본자료형
- int : 정수
- float : 실수
- str : 문자열
- bool : 논리형
2. 자료구조
- list : [] / list()
- tuple : () / tuple()
- dict : {} / dict()
- set: : {} / set()

```python
data_set = set()
```

```python
print(type(data_set))
```

```python
ppap = set()
print(type(ppap))
```

```python
ppap = {"pen", "apple", "pineapple", "pen"} # set은 중복값을 허용하지 않음!
```

```python
print(ppap)
```

```python
# set 이라는 자료구조 내 특정 값이 있는지 여부를 판단하는 문법 존재
# set은 애초에 반복적인 값을 허용 x, set 자료구조에 이미 저장된 요소들은 모두 1번씩 밖에 사용

"apple" in ppap # in 전치사의 역할 : in 뒤에 나오는 set 자료구조 안에 값이 한번이라도 있으면 true)
"applepen" in ppap
```

```python
data1 = {"apple", "samsung", "lg"}
data2 = {"samsung", "lg", "xiaomi"}
```

```python
data1 & data2 # 교집합
```

```python
data1 | data2 # vertical bar # 합집합
```

```python
data1 - data2
```

```python
data1 ^ data2
```
