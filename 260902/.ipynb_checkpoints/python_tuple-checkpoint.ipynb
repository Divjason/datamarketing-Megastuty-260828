```python
# parameter : 매개변수
# argument : 인수 = 인자값

def mul_return(a) :
    b = a + 1
    return a, b # 함수의 반환문, return은 반환 시, 값의 타입을 tuple
```

```python
result = mul_return(1)
print(result, type(result))
```

```python
len(result) # len = length 약어 = 길이 = tuple 자료구조 내 아이템 갯수
print(result[0]) # tuple, list => index를 가지고 있음
```

```python
# del result[3]

first, second = result # 구조분해할당 = unpacking
```

```python
print(first, second)
```
