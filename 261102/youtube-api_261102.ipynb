```python
import os
from dotenv import load_dotenv

load_dotenv()

API_KEY = os.getenv("YOUTUBE_API_KEY")
```

```python
from googleapiclient.discovery import build

youtube = build(
    "youtube",
    "v3",
    developerKey=API_KEY
)

print("Youtube API 서비스 객체 생성 성공!")
```

youtube 객체 : 유튜브 내 컨텐츠를 받아올 수 있도록 하는 중간역할
- search() : 유튜브 채널 및 영상 관련 정보
- videos() : 영상 관련 상세 통계정보 (좋아요, 댓글수)
- channels() : 영상 채널정보
- playlists() : 재생목록 관련 정보
- commentThreads() : 영상 관련 댓글 정보

```python
# 영상 정보를 가져올 때, 특정 조건에 맞춰서 가져오고 싶다면?
# list() : 검색 필터링 설정 (메서드 체이닝)
# 영상제목, 영상설명, 영상채널명, 영상게재된시점, 썸네일 정보 -> 
request = youtube.search().list(
    part="snippet",
    q="치이카와",
    type="video",
    maxResults=10,
    order="viewCount",
    regionCode="KR",
    relevanceLanguage="ko"
)

response = request.execute() # response -> 요청 값이 들어옴!!!
```

```python
response
```

```python
import pandas as pd

videos = list()

for item in response["items"] :
    snippet = item["snippet"]

    videos.append({
        "video_id": item["id"]["videoId"],
        "title": snippet["title"],
        "channel": snippet["channelTitle"],
        "published_at": snippet["publishedAt"],
        "description": snippet["description"]
    })

df = pd.DataFrame(videos)

df
```
