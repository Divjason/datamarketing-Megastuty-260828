```python
import json
import openpyxl

from openpyxl.styles import Alignment, Font, PatternFill # 엑셀파일 내 문서텍스트 정렬, 폰트, 배경컬러 등
from openpyxl.drawing.image import Image as ExcelImage
from io import BytesIO

with open("response.json", "r", encoding="utf-8") as f :
    data = json.load(f)

wb = openpyxl.Workbook() # openpyxl 모듈 활용, 신규 엑셀파일 생성
ws = wb.active # 생성된 엑셀파일 내 시트생성 후 선택
ws.title = "캠핑장정보"
headers = ["아이디", "캠핑장명", "이미지", "지역", "주소", "홈페이지"]
ws.append(headers)

# 삽입하고자 하는 이미지 픽셀 사이즈
# 엑셀은 너비 : 텍스트 문자 갯수 : 1개의 텍스트 문자 차지하는 7 : 12, 좌.우측 각각 여백(마진.패딩) : 7
IMG_W_PX = 100 # 사이즈 단위 : px 가정
IMG_H_PX = 75 # 사이즈 단위 : px 가정

# 우리가 원하는 픽셀 단위의 사이즈를 엑셀시트에 맞춰서 단위 변환시켜주는 함수 정의
def px_to_col_width(px) :
    return (px - 14) / 7

def px_to_row_height(px) :
    return px * 0.75

COL_W_TF = px_to_col_width(IMG_W_PX)
ROW_H_TF = px_to_row_height(IMG_H_PX)

# 엑셀 스타일 정의 영역
cell_align = Alignment(horizontal="left", vertical="center", wrap_text=True)
header_fill = PatternFill("solid", fgColor="F2F2F2")
header_font = Font(bold=True)

for col in range(1, len(headers) + 1) :
    c = ws.cell(row=1, column=col)
    c.alignment = Alignment(horizontal = "center", vertical  = "center")
    c.fill = header_fill
    c.font = header_font

# first = data["response"]
# second = first["body"]
# third = second["items"]
items = data["response"]["body"]["items"]["item"]
row = 2

for item in items :
    contentId = item.get("contentId", "")
    facltNm = item.get("facltNm", "")
    firstImageUrl = item.get("firstImageUrl", "")
    doNm = item.get("doNm", "")
    addr1 = item.get("addr1", "")
    homepage = item.get("homepage", "")

    ws.append([contentId, facltNm, "", doNm, addr1, homepage])
    ws.row_dimensions[row].height = ROW_H_TF

    for col in range(1, 7) :
        ws.cell(row=row, column=col).alignment = cell_align

    if firstImageUrl :
        try :
            img_data = requests.get(firstImageUrl).content
            img_file = BytesIO(img_data)
            img = ExcelImage(img_file)
            img.width = IMG_W_PX
            img.height = IMG_H_PX
            ws.add_image(img, f"C{row}")
        except Exception as e :
            print(f"이미지 다운로드 실패 : {firstImageUrl} -> {e}")

    row += 1

ws.column_dimensions["A"].width = 12
ws.column_dimensions["B"].width = 30
ws.column_dimensions["C"].width = COL_W_TF
ws.column_dimensions["D"].width = 15
ws.column_dimensions["E"].width = 30
ws.column_dimensions["F"].width = 50

file_name = "camping_info.xlsx"

ws.freeze_panes = "A2"

wb.save(file_name)

print(f"✅ {file_name} 파일 저장 완료")
```
